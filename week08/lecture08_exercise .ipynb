```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import json
import requests
```

```python
gm = px.data.gapminder()
gm_2007 = gm[gm["year"] == 2007].copy()

global_avg_lifeexp = gm_2007["lifeExp"].mean()
gm_2007["deviation"] = gm_2007["lifeExp"] - global_avg_lifeexp

fig = px.choropleth(
    data_frame=gm_2007,
    locations='country',
    locationmode='country names',
    color='deviation',
    color_continuous_scale=px.colors.diverging.RdBu,
    color_continuous_midpoint=0,
    fitbounds='locations',
    hover_name='country',
    hover_data={'country': False, 'lifeExp': ':.2f', 'deviation': ':.2f'},
    labels={'deviation': 'Deviation from Global Average'},
    title='2007 Life Expectancy: African regions are furthest below the global average',
    height=500
)

fig.update_layout(
    font=dict(family='Arial', size=12),
    geo=dict(showframe=False, showcoastlines=True, coastlinecolor="black"),
    margin=dict(l=0, r=0, t=55, b=0)
)
fig.show()
```

Task1 Answer： Chart type: px.choropleth Reason: Designed for global country boundary maps. Simple country outlines are ideal for cross-country comparison. Colour scale: Diverging scale Reason: Data shows positive/negative deviation from a global mean. A diverging scale clearly distinguishes values above and below the average, with zero as the neutral midpoint.

```python
# Step 1: Build dataset (World countries + Life Expectancy 2007)
# Extract country name and life expectancy for matching
world_data = gm_2007[["country", "lifeExp"]].copy()
world_data.rename(columns={"country": "Country"}, inplace=True)

# --------------------------
# Step 2: Load your assigned GeoJSON file
# --------------------------
geojson_raw_url = "https://raw.githubusercontent.com/datasets/geo-countries/main/data/countries.geojson"
response = requests.get(geojson_raw_url)
world_geojson = json.loads(response.text)

#  Inspect GeoJSON properties
print("=== GeoJSON Properties Check ===")
print(world_geojson['features'][0]['properties'])

# --------------------------
# Step 2: px.choropleth_map 
# --------------------------
fig2 = px.choropleth_map(
    data_frame=world_data,
    geojson=world_geojson,
    locations='iso_alpha',
    featureidkey='properties.ISO3166-1-Alpha-3',
    color='lifeExp',
    color_continuous_scale='Oranges',
    hover_name='country',
    hover_data={'country': False, 'iso_alpha': False},
    labels={'lifeExp': 'Life Expectancy (2007)'},
    title='Global Life Expectancy Distribution (2007)',
    center={'lat': 20.0, 'lon': 0.0},
    zoom=1,
    opacity=0.7,
    height=700
)

fig2.update_layout(
    font=dict(family='Arial', size=12),
    margin=dict(l=0, r=0, t=55, b=0)
)

fig2.show()
```

Task2 Answer
GeoJSON source: https://github.com/datasets/geo-countries/blob/main/data/countries.geojson Chart type chosen: px.choropleth: Clean outline map for direct cross-country comparison. px.choropleth_map: Interactive basemap with geographic context. Colour scale chosen: Sequential (Oranges) — Life expectancy is a continuous non-negative value, so sequential scale clearly shows magnitude differences.
